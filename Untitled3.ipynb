# 3.Data Wrangling 
  The dataset was cleaned to remove irregular or unnecessary records that could affect the analysis. Missing values were checked and handled appropriately. Weekday records with birth counts less than 10,000 were removed to eliminate abnormal observations. This ensured the dataset was consistent and suitable for statistical analysis and visualization.

```python
#Check Unique Days
df['day_of_week'].unique()

#1=Monday,2=Tuesday 
```

```python
#Remove Weekday Rows With Births < 10000
df = df[~((df['day_of_week'].isin([1,2,3,4,5])) & (df['births'] < 10000))]
```

# 4.Data Visualization 
Box Plot:
Box and whisker plots were used to show the distribution of births across different days of the week. These plots help identify variability and potential outliers.

Histogram:
Histograms were used to show the frequency distribution of daily birth counts.

KDE Plot:
Kernel Density Estimation (KDE) plots were used to visualize the density distribution of birth counts.

```python
#Box Plot
plt.figure(figsize=(8,5))
sns.boxplot(x='day_of_week', y='births', data=df)
plt.title("Birth Distribution by Day of Week")
plt.show()
```

```python
#Histogram
plt.figure(figsize=(8,5))
sns.histplot(df['births'], kde=True)
plt.title("Histogram of Birth Counts")
plt.show()
```

```python
#KDE Plot
plt.figure(figsize=(8,5))
sns.kdeplot(df['births'], fill=True)
plt.title("KDE Plot of Birth Distribution")
plt.show()
```

# 5.Statistical Analysis
Average Births:
The average births for each day of the week were calculated.

Median Births;
The median value was calculated to find the middle birth count.

Dispersion:
Standard deviation was used to measure variation in birth counts.

Interquartile Range (IQR):
IQR was calculated to understand the spread of the middle 50% of data.

Outliers:
Outliers were detected using the IQR method to find unusually high or low birth counts.

```python
#Average Births Per Day of Week
avg_births = df.groupby('day_of_week')['births'].mean()
print(avg_births)
```

```python
#Median Births
median_births = df.groupby('day_of_week')['births'].median()
print(median_births)
```

```python
#Dispersion (Standard Deviation)
std_births = df.groupby('day_of_week')['births'].std()
print(std_births)
```

```python
#Interquartile Range (IQR)
Q1 = df['births'].quantile(0.25)
Q3 = df['births'].quantile(0.75)

IQR = Q3 - Q1
print("IQR:", IQR)
```

```python
#Finding the Outliers
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['births'] < lower) | (df['births'] > upper)]

print(outliers)
```

# 6.Analysis of Birth Distribution Across the Year
50% Birth Occurrence per Year:

The cumulative number of births was calculated for each year. The analysis identified the day of the year when 50% of the total births had occurred.
The results showed that in most years between 2000 and 2014, the 50% birth mark occurs around early July.

50% Birth Period in the Year 2000:

A separate analysis was performed for the year 2000 to determine the point when half of the births had occurred. The analysis showed that approximately half of the births occurred by early July in 2000.

```python
#Find Day When 50% Births Occur Each Year
df['cumulative_births'] = df.groupby('year')['births'].cumsum()

df['total_births'] = df.groupby('year')['births'].transform('sum')

df['percentage'] = df['cumulative_births'] / df['total_births']

result = df[df['percentage'] >= 0.5].groupby('year').first()

print(result)
```

```python
#50% Birth Period in Year 2000
data_2000 = df[df['year'] == 2000].copy()

data_2000['cumulative'] = data_2000['births'].cumsum()

total = data_2000['births'].sum()

data_2000['percentage'] = data_2000['cumulative'] / total

data_2000[data_2000['percentage'] >= 0.5].head(1)
```
